# C-MAPSS exploratory data analysis

Explore the raw NASA C-MAPSS FD001 run-to-failure data. The repository stores this dataset through DVC; run `dvc pull` from the repository root if the file is missing.


```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
while ROOT != ROOT.parent and not (ROOT / "data").exists():
    ROOT = ROOT.parent
raw_path = ROOT / "data" / "raw" / "train_FD001.txt"
if not raw_path.exists():
    raise FileNotFoundError(f"{raw_path} is missing. Run `dvc pull` from the repository root.")

columns = ["unit", "cycle", "setting_1", "setting_2", "setting_3"] + [f"sensor_{i}" for i in range(1, 22)]
data = pd.read_csv(raw_path, sep=r"\s+", header=None, names=columns)
data.head()
```

```python
fleet = data.groupby("unit").agg(
    cycles=("cycle", "max"),
    first_cycle=("cycle", "min"),
)
fleet.describe().round(1)
```

```python
sensor_columns = [column for column in data.columns if column.startswith("sensor_")]
data[sensor_columns].describe().T.head(10)
```

```python
unit_id = int(data["unit"].min())
unit_data = data.loc[data["unit"] == unit_id]
plt.figure(figsize=(10, 4))
plt.plot(unit_data["cycle"], unit_data["sensor_2"], linewidth=1.5)
plt.xlabel("Cycle")
plt.ylabel("Sensor 2 reading")
plt.title(f"C-MAPSS FD001: sensor 2 trajectory for unit {unit_id}")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```
